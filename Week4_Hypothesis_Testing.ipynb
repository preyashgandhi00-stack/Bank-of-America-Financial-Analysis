```python
import pandas as pd
from scipy import stats

df = pd.read_csv("Bank_of_America_Cleaned.csv")

df["Date"] = pd.to_datetime(df["Date"], utc=True)
df = df.sort_values("Date").reset_index(drop=True)

print("Rows and columns:", df.shape)
print("Columns:", df.columns.tolist())
df.head()
```

```python
df["Daily_Return"] = df["Close"].pct_change(fill_method=None)
returns = df["Daily_Return"].dropna()

print("Observations:", len(returns))
print("Mean daily return:", f"{returns.mean():.6%}")
print("Median daily return:", f"{returns.median():.6%}")
print("Standard deviation:", f"{returns.std():.6%}")
```

```python
test_result = stats.ttest_1samp(returns, popmean=0)

print("t-statistic:", test_result.statistic)
print("p-value:", test_result.pvalue)
print("Degrees of freedom:", len(returns) - 1)
```

```python
import numpy as np

n = len(returns)
mean_return = returns.mean()
standard_error = returns.std(ddof=1) / np.sqrt(n)

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=n - 1,
    loc=mean_return,
    scale=standard_error
)

print("95% confidence interval:", confidence_interval)
```

## Hypothesis Test: Is the Mean Daily Return Different from Zero?

**Null hypothesis (H₀):** The mean daily return is zero.  
**Alternative hypothesis (H₁):** The mean daily return is different from zero.

A two-sided one-sample t-test was performed at the 5% significance level (α = 0.05) using 13,328 daily returns. The mean daily return was 0.054831%. The test produced a t-statistic of 2.6783 and a p-value of 0.0074.

Since the p-value is below 0.05, the null hypothesis is rejected. The 95% confidence interval for the mean daily return was approximately 0.0147% to 0.0950%, which does not include zero. This suggests that the historical mean daily return in this dataset was positive and statistically different from zero.

**Limitation:** This test uses historical closing-price returns. It does not predict future performance, and the test assumes observations are independent. Daily financial returns can be related over time, so the result should be interpreted with caution.
